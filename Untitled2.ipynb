```python
import pandas as pd

df = pd.read_excel(
    "/content/Global Carbon clean data.xlsx",
    sheet_name="Table1"
)

print(df.shape)
print(df.columns.tolist())
```

```python
df.isnull().sum()
```

```python
null_percentage = (df.isnull().sum() / len(df)) * 100

print(
    null_percentage[null_percentage > 0]
    .sort_values(ascending=False)
)
```

```python
main_columns = [
    'population',
    'gdp',
    'co2',
    'cement_co2',
    'coal_co2',
    'oil_co2'
]

for col in main_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = (
        (df[col] < lower) |
        (df[col] > upper)
    ).sum()

    print(col, "→", outliers, "outliers")
```

```python
main_columns = [
    'population',
    'gdp',
    'co2',
    'cement_co2',
    'coal_co2',
    'oil_co2'
]

for col in main_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(
        lower=lower,
        upper=upper
    )
```

```python
for col in main_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = (
        (df[col] < lower) |
        (df[col] > upper)
    ).sum()

    print(col, "→", outliers, "outliers")
```

```python
print("Shape:", df.shape)

print("\nNull Values:")
print(df.isnull().sum())

print("\nDuplicates:", df.duplicated().sum())
```
